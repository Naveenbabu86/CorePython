loop

dictionary access

This is a very common Python interview question.

```python
employees = [
    {"name":"Naveen","salary":50000},
    {"name":"Rahul","salary":80000},
    {"name":"Priya","salary":60000}
]
```

```python
employees
```

```python
max_sal = 0

for emp in employees:
    for key, value in emp.items():

        if type(value) == int:

            if value > max_sal:
                max_sal = value


print(max_sal)
```

Salary < 50000      -> Low
50000 - 79999       -> Medium
>= 80000            -> High

```python
employees = [
    {"name":"Naveen","salary":50000},
    {"name":"Rahul","salary":80000},
    {"name":"Priya","salary":60000},
    {"name":"Amit","salary":90000},
    {"name":"Anjali","salary":45000}
]
```

```python
Low_count = 0
Medium_count = 0
High_count = 0
for emp_data in employees:
    for emp_name , emp_sal in emp_data.items():
        if type(emp_sal) == int:
            if emp_sal<50000:
                Low_count+=1
                
            elif emp_sal>=50000 and emp_sal<=79999:
                Medium_count+=1
                
            elif emp_sal>=80000:
                High_count+=1
                
print(f"Low_count is {Low_count}")
print(f"Medium_count is {Medium_count}")
print(f"High_count is {High_count}")
            
```

numbers = [10, 20, 30, 20, 40, 50, 10, 60]
Task

Print only the duplicate numbers.

Expected Output:

10
20

```python
numbers = [10, 20, 30, 20, 40, 50, 10, 60]
l= []
duplicate_list = []

for i in numbers:
    if i not in l:
        l.append(i)
    else:
        duplicate_list.append(i)

print(duplicate_list)
```
