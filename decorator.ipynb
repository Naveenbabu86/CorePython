```python
# What is a Decorator?

# A decorator is a function that adds extra behavior to another function without modifying its original code.

# Think of it like a wrapper.

# Original Function
#        ↓
# Decorator wraps it
#        ↓
# Enhanced Function
```

```python
class Math:

    @staticmethod
    def add(a, b):
        return a + b
```

```python
m = Math()
m.add(1,2)
```
