```python

# encapsulation : building data and methods together is calle encapsulation

class student1():
    def __init__(self,name: str):
        self.name = name
    def display(self):
        return f"my name is {self.name}"
 
    
    
```

```python
# inheritence

class pearson:
    def speak(self):
        print("speaking")
class student(pearson):
    pass
```

```python
s = student()
print(s.speak())
```

```python
# polymorphisam , same method different behaviour
class dog():
    def sound(self):
        return "barking"
class cat():
    def sound(self):
        return "meow"
```

```python


animals = [dog(),cat()]
for animal in animals:
    print(animal.sound())
```

```python
# Abstraction , Abstraction is the process of hiding implementation details and exposing only the essential functionality to the user.
"""
Abstraction is an OOP principle that hides complex implementation details and exposes only the necessary features through a simple interface. 
In Python, abstraction is typically implemented using abstract classes and abstract methods from the abc module.
"""

from abc import ABC, abstractmethod

class Agent(ABC):

    @abstractmethod
    def execute(self):
        pass
```

```python
class IncidentAgent(Agent):

    def execute(self):
        print("Analyzing incident")
```

```python
class NotificationAgent(Agent):

    def execute(self):
        print("Sending notification")
```

```python
i = IncidentAgent()
i.execute()
```
